```python
df_copy['Size']=df_copy['Size'].str.replace('M','000')
df_copy['Size']=df_copy['Size'].str.replace('k','')
df_copy['Size']=df_copy['Size'].replace('Varies with device',np.nan)
df_copy['Size']=df_copy['Size'].astype(float)
```

```python
df_copy.info()
```

```python
df_copy['Size']
```

```python
df_copy['Installs'].unique()
```

```python
df_copy['Price'].unique()
```

```python
chars_to_remove=['+',',','$']
cols_to_clean=['Installs','Price']
for item in chars_to_remove:
    for cols in cols_to_clean:
        df_copy[cols]=df_copy[cols].str.replace(item,'')
```

```python
df_copy['Price'].unique()
```

```python
df_copy['Installs'].unique()
```

```python
df_copy['Installs']=df_copy['Installs'].astype('int')
df_copy['Price']=df_copy['Price'].astype('float')
```

```python
df_copy.info()
```

```python
## Handling Last Updates feature
df_copy['Last Updated'].unique()
```

```python
df_copy['Last Updated']=pd.to_datetime(df_copy['Last Updated'])
df_copy['Day']=df_copy['Last Updated'].dt.day
df_copy['Month']=df_copy['Last Updated'].dt.month
df_copy['Year']=df_copy['Last Updated'].dt.year
```

## Which category has largest number of Installations

```python
df_cat_installs = df_copy.groupby(['Category'])['Installs'].sum().sort_values(ascending = False).reset_index()
df_cat_installs.Installs = df_cat_installs.Installs/1000000000# converting into billions
df2 = df_cat_installs.head(10)
plt.figure(figsize = (14,10))
sns.set_context("talk")
sns.set_style("darkgrid")

ax = sns.barplot(x = 'Installs' , y = 'Category' , data = df2,palette='rocket' )
ax.set_xlabel('No. of Installations in Billions')
ax.set_ylabel('')
ax.set_title("Most Popular Categories in Play Store", size = 20)
plt.show()
```

## Insights 
1. Out of all the categories "GAME" has the most number of Installations
2. With almost 14 Billion Installations Game is the most popular Category in Google App Store

## What are the Top 5 most installed Apps in each popular Category 

```python
import matplotlib.pyplot as plt
import seaborn as sns

# prepare data
dfa = df_copy.groupby(['Category', 'App'])['Installs'].sum().reset_index()
dfa = dfa.sort_values('Installs', ascending=False)

apps = ['GAME', 'COMMUNICATION', 'PRODUCTIVITY', 'SOCIAL']
palettes = ['rocket', 'crest', 'mako', 'flare']

# convert installs to billions for readability
dfa['Installs_B'] = dfa['Installs'] / 1e9

sns.set_context("poster")
sns.set_style("darkgrid")

plt.figure(figsize=(20,16))   # 2x2 layout so smaller width than before

for i, (app, pal) in enumerate(zip(apps, palettes)):
    df2 = dfa[dfa.Category == app]
    df3 = df2.head(5)
    
    ax = plt.subplot(2, 2, i + 1)      # 2 rows x 2 cols for 4 plots
    sns.barplot(data=df3, x='Installs_B', y='App', palette=pal, ax=ax)
    
    ax.set_xlabel('Installations (Billions)', fontsize=16)
    ax.set_ylabel('', fontsize=14)
    ax.set_title(app, size=20, weight='bold')
    
    # annotate bar values
    for j, v in enumerate(df3['Installs_B']):
        ax.text(v + 0.02, j, f'{v:.2f}B', va='center', fontsize=14)

plt.tight_layout()
plt.subplots_adjust(hspace=0.3)
plt.show()
```

## Insights 
1. Most popular Game is Subway Surfer
2. Most popular Communication app is Skype
3. Most popular Productivity app is Google Drive
4. Most popular Socila app is Facebook

## Free vs Paid Apps

```python
sns.boxplot(x='Type', y='Rating', data=df_copy)
plt.title("Free vs Paid Apps: Rating Comparison")
plt.show()
```

## Insights 
Both free and paid apps receive similar median ratings (around 4.3). However, free apps exhibit a wider range of ratings with several low-rated outliers, indicating inconsistency in quality. Paid apps, though fewer, maintain relatively stable and slightly higher ratings on average.

## Rating and Installs Coorelation 

```python
sns.scatterplot(x='Installs', y='Rating', data=df_copy, alpha=0.5)
plt.xscale('log')
plt.show()
```

## Insights 
There is no strong correlation between installs and ratings. While popular apps (with millions of installs) generally maintain good ratings above 4, the overall pattern suggests that a higher number of installs does not guarantee better user ratings. Less popular apps show more variation in ratings, including several low-rated ones.

## Category Wise Average Rating 

```python
avg_rating = df_copy.groupby('Category')['Rating'].mean().sort_values(ascending=False).reset_index()
sns.barplot(x='Rating', y='Category', data=avg_rating.head(10), palette='mako')
plt.title("Top 10 Categories by Average Rating")
plt.show()
```

## Insight 
Education and creative-focused apps receive the best user feedback, while entertainment-based apps, despite their popularity, show slightly lower satisfaction levels.
